#Datanın vizual olaraq təsviri

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
energy_cols=["appliances","lights"]

temp_cols=[
    "t1","t2","t3","t4","t5",
    "t6","t7","t8","t9"
]
humidity_cols=[
    "rh_1","rh_2","rh_3","rh_4","rh_5",
    "rh_6","rh_7","rh_8","rh_9"
]

weather_cols=[
    "t_out","press_mm_hg","rh_out",
    "windspeed","visibility","tdewpoint"
]
```

```python
df[energy_cols]["appliances"].describe()
```

```python
plt.figure(figsize=(10,5))
sns.histplot(data=df, x="appliances",bins=30,kde=True)

plt.title("Enerji serfiyyati histoqram analizi")
plt.xlabel("Enerji Sərfiyyatı")
plt.ylabel("Sərfiyyat sayı")
plt.show()
```

```python
plt.figure(figsize=(3,10))

sns.boxplot(data=df,y="appliances")

plt.title("Enerji serfiyyati Box(outlier) analizi")
plt.show()
```

```python
df[energy_cols]["lights"].describe()
```

```python
df["lights"].value_counts().sort_index()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data=df,x="lights")

plt.title("Sensorların işıq sərfiyyatı")
plt.xlabel("Lights")
plt.ylabel("Müşahidə sayı / tezliyi")

plt.show()
```

#Temperatur sensorlari analizi

```python
df[temp_cols].describe().T
```

```python
mean=df[temp_cols].mean()
plt.figure(figsize=(12,5))

sns.barplot(x=mean.index,y=mean.values)
plt.title("Otaqlar üzrə orta temperatur müqayisəsi")
plt.xlabel("Temperatur Sensorları")
plt.ylabel("Orta Temperatur")
plt.show()
```

Date ile sutunlarin elaqelendirilmesi

```python
df["hour"]=df["date"].dt.hour
df["day_name"]=df["date"].dt.day_name()
df["month"]=df["date"].dt.month
df["year"]=df["date"].dt.year
```

#saatlara gore enerjinin serfiyyati

```python
df.head()
```

```python
hourly_energy=df.groupby("hour")[energy_cols].mean()
hourly_energy.head()
```

```python
hourly_energy.plot(subplots=True,figsize=(12,8),marker="o")

plt.title("Saatlara görə Enerji Sərfiyyatı və Lights")
plt.xlabel("Saatlar 0-23")
plt.ylabel("Orta dəyər")

plt.show()
```

#Heftesonu ve hefteici sutunlarin muqayisesi

```python
df["dayofweek"]=df["date"].dt.day_of_week
df["is_weekend"]=df["dayofweek"]>=5
```

```python
hourly_week=df.groupby(["is_weekend","hour"])[energy_cols].mean().unstack(level=0)
```

```python
fig, axes=plt.subplots(1,2,figsize=(14,5))


axes[0].plot(hourly_week.index,hourly_week["appliances"][False],marker="o",label="İş günləri",color="red")
axes[0].plot(hourly_week.index,hourly_week["appliances"][True],marker="o",label="Həftəsonu günləri",color="blue")
axes[0].set_title("Appliance-İş günləri vs Həftəsonu")
axes[0].set_xlabel("Saat (0-23)")
axes[0].set_ylabel("Orta Sərfiyyat")
axes[0].legend()
axes[0].grid()


axes[1].plot(hourly_week.index,hourly_week["lights"][False],label=("İş günləri"),color="red")
axes[1].plot(hourly_week.index,hourly_week["lights"][True],label=("Həftəsonu günləri"),color="blue")
axes[1].set_title("Lights-iş günləri vs Həftəsonu")
axes[1].set_xlabel("Saat (0-23)")
axes[1].set_ylabel("Orta dəyər")
axes[1].legend()
axes[1].grid()
```

```python
df.groupby("is_weekend")[energy_cols].agg(["mean","median"])
```

```python
monthly_energy=df.groupby("month")[energy_cols].mean()

monthly_energy.head()
```

```python
plt.figure(figsize=(10,5))

plt.plot(monthly_energy.index,monthly_energy["appliances"])
plt.xlabel("Aylar")
plt.ylabel("Orta enerji serfiyyatı")

plt.xticks([1,2,3,4,5])
plt.show()
```

```python
df[weather_cols].describe().T
```

```python
corr_matrix=df[energy_cols + weather_cols].corr()

plt.figure(figsize=(10,6))
sns.heatmap(corr_matrix,annot=True,cmap="coolwarm",fmt=".2f",linewidths=0.5)
plt.title("Enerji və Hava Parametrləri arasındakı Korrelyasiya")

plt.show()
```

```python
df[humidity_cols].describe().T
```

```python
corr_uclu=df[energy_cols + weather_cols + humidity_cols].corr()

plt.figure(figsize=(12,6))
sns.heatmap(data=corr_uclu,annot=True,fmt=".2f",cmap="coolwarm")
plt.title("Enerji sərfiyyatı,Hava Parametrləri və Rütubətlilik arasında əlaqə")

plt.show()
```
